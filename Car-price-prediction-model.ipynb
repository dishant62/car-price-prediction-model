```python
X= df.drop(columns=['price'])
y= df['price']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test= train_test_split(X, y, test_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline
```

```python
ohe= OneHotEncoder()
ohe.fit(X[['name', 'company', 'fuel_type']])
```

```python
column_trans= make_column_transformer((OneHotEncoder(categories= ohe.categories_),['name', 'company', 'fuel_type']), remainder='passthrough' )
```

```python
lr= LinearRegression()
```

```python
pipe= make_pipeline(column_trans, lr)
```

```python
pipe.fit(X_train, y_train)
```

```python
y_pred = pipe.predict(X_test)
```

```python

r2_score(y_test, y_pred)
```

```python
scores=[]
for i in range(1000):
    X_train, X_test, y_train, y_test= train_test_split(X, y, test_size=0.2, random_state=i)
    lr= LinearRegression()
    pipe = make_pipeline(column_trans, lr)
    pipe.fit(X_train, y_train)
    y_pred= pipe.predict(X_test)
    scores.append(r2_score(y_test, y_pred))
```

```python
np.argmax(scores)
```

```python
scores[np.argmax(scores)]
```

```python
X_train, X_test, y_train, y_test= train_test_split(X, y, test_size=0.2, random_state=np.argmax(scores))
lr= LinearRegression()
pipe = make_pipeline(column_trans, lr)
pipe.fit(X_train, y_train)
y_pred= pipe.predict(X_test)
r2_score(y_test, y_pred)
```

```python
import gradio as gr
import pandas as pd

def predict_price(car_name, company, year, kms_driven, fuel_type):
    # Create a DataFrame from user input
    user_car_data = pd.DataFrame([[car_name, company, year, kms_driven, fuel_type]],
                                 columns=['name', 'company', 'year', 'kms_driven', 'fuel_type'])

    # Predict the price using the trained pipeline
    predicted_price = pipe.predict(user_car_data)[0]
    return f"Predicted car price: ₹{predicted_price:,.2f}"

# Get unique values for dropdowns from the 'df' DataFrame
car_names = df['name'].unique().tolist()
companies = df['company'].unique().tolist()
fuel_types = df['fuel_type'].unique().tolist()

# Create Gradio interface
iface = gr.Interface(
    fn=predict_price,
    inputs=[
        gr.Dropdown(choices=car_names, label="Car Name"),
        gr.Dropdown(choices=companies, label="Company"),
        gr.Number(label="Manufacturing Year (e.g., 2015)", precision=0),
        gr.Number(label="Kilometers Driven (e.g., 60000)", precision=0),
        gr.Dropdown(choices=fuel_types, label="Fuel Type")
    ],
    outputs="text",
    title="Car Price Predictor",
    description="Enter car details to get an estimated price."
)

# Launch the interface
iface.launch()
```
